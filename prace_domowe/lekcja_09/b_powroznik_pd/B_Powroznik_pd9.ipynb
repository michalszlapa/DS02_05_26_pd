Figura z metodą pole().
Stwórz klasy potomne:
Prostokat(szerokosc, wysokosc)
Kolo(promien)
Każda klasa implementuje
pole().
Stwórz listę figur i oblicz łączne pole.

```python
import math

class Figura:
    def pole(self):
        """Metoda bazowa, która będzie nadpisywana w klasach potomnych."""
        raise NotImplementedError("Podklasa musi zaimplementować tę metodę")

class Prostokat(Figura):
    def __init__(self, szerokosc, wysokosc):
        self.szerokosc = szerokosc
        self.wysokosc = wysokosc

    def pole(self):
        return self.szerokosc * self.wysokosc

class Kolo(Figura):
    def __init__(self, promien):
        self.promien = promien

    def pole(self):
        return math.pi * (self.promien ** 2)

# Tworzenie listy figur
figury = [
    Prostokat(5, 10),  # Pole: 50
    Kolo(3),           # Pole: ~28.27
    Prostokat(2, 4)    # Pole: 8
]

# Obliczanie łącznego pola
laczne_pole = sum(figura.pole() for figura in figury)

print(f"Łączne pole wszystkich figur wynosi: {laczne_pole:.2f}")
```
